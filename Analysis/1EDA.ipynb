```python
# import python essential libraries
import ast
import pandas as pd
import seaborn as sb
from datasets import load_dataset
import matplotlib.pyplot as plt

# load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# data_cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['data_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
job_count = df['job_title_short'].value_counts().to_frame()

sb.set_theme(style='ticks')
sb.barplot(data=job_count, x='count', y='job_title_short', hue='count', palette='dark:b_r', legend=True)
sb.despine()
plt.title('Number of jobs available for data science role')
plt.xlabel('number of jobs')
plt.ylabel('')
plt.show()
```

```python
loc_count = df['job_country'].value_counts().to_frame().head(20)

sb.set_theme(style='ticks')
sb.barplot(data=loc_count, x='count', y='job_country', hue='count', palette='dark:b_r', legend=True)
sb.despine()
plt.title('Number of jobs available for top 20 countries')
plt.xlabel('number of jobs')
plt.ylabel('')
plt.show()
```

```python
company_data = df['company_name'].value_counts().to_frame()[1:].head(20)

sb.set_theme(style='ticks')
sb.barplot(data=company_data, x='count', y='company_name', hue='count', palette='dark:b_r', legend=True)
sb.despine()
plt.title('Number of jobs available for top 20 companies')
plt.xlabel('number of jobs')
plt.ylabel('')
plt.show()
```

```python
service_offers = {
    "job_work_from_home":"Work From Home Offered",
    "job_no_degree_mention":"Degree Requirement?",
    "job_health_insurance":"Health Insurance"
}

fig, ax = plt.subplots(1, 3, figsize=(11,3.5))

for i, (column, title) in enumerate(service_offers.items()):
    ax[i].pie(df[column].value_counts(), labels=['False', 'True'], startangle=90)
    ax[i].set_title(title)

plt.show()
```
